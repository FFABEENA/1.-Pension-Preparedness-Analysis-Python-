## 1. Problem Statement

### The Problem

Pension subscribers may differ considerably in their ability to prepare financially for retirement. Differences in income, contribution levels, investment allocation, accumulated pension fund value, retirement age, and expected pension income may influence the level of financial preparedness among subscribers.

However, meaningful analysis of these factors requires reliable and consistent data. Pension-related datasets may contain data-quality issues such as missing values, duplicate records, inconsistent categories, incorrect data types, invalid numerical values, and logical inconsistencies between related variables. If these issues are not identified and addressed, they may affect the accuracy and reliability of the analysis.

### What This Project Aims to Solve

This project aims to address two related problems:

1. **Data reliability problem:**  
   How can the current pension subscriber dataset be explored, cleaned, validated, and transformed into a reliable dataset suitable for further analysis?

2. **Pension preparedness problem:**  
   After obtaining a reliable dataset, what demographic, employment, contribution, investment, and retirement-related factors are associated with stronger or weaker pension preparedness among subscribers?

The analysis will use exploratory data analysis, statistical summaries, and visualizations to identify patterns, relationships, differences, and potential areas of financial vulnerability within the dataset.

### 6.9.35 Outlier Detection Using Boxplots

Boxplots were used to visually examine the distribution of the numerical variables and identify potential outliers. The boxplot displays the median, first quartile (Q1), third quartile (Q3), and the interquartile range (IQR).

Observations falling below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are displayed as individual points and are considered potential outliers.

The boxplots were used as a preliminary diagnostic rather than as an automatic basis for removing observations. Extreme values may represent genuine differences between individuals in income, contributions, pension fund value, NAV, units held, or other financial variables. Therefore, potential outliers were reviewed in the context of the dataset and were not removed solely because they appeared outside the boxplot whiskers.

```python
# 6.9.35 Outlier Detection Using Boxplots

import matplotlib.pyplot as plt
import seaborn as sns

boxplot_columns = [
    "Annual_Income_INR",
    "Monthly_Contribution_INR",
    "Employer_Monthly_Contribution_INR",
    "Total_Contribution_Till_Date_INR",
    "Current_Fund_Value_INR",
    "Expected_Monthly_Pension_INR",
    "Current_NAV_INR",
    "Total_Units_Held",
    "Years_of_Contribution",
    "Years_to_Retirement"
]

fig, axes = plt.subplots(2, 5, figsize=(20, 11))

colors = sns.color_palette("Set2", 10)

for ax, column, color in zip(axes.flatten(), boxplot_columns, colors):

    sns.boxplot(
        y=df[column],
        ax=ax,
        color=color,
        width=0.55
    )

    ax.set_title(
        column.replace("_", " "),
        fontsize=11,
        fontweight="bold"
    )

    ax.set_xlabel("")
    ax.set_ylabel(column.replace("_", " "), fontsize=9)

    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.35
    )

fig.suptitle(
    "Outlier Detection Using Boxplots",
    fontsize=18,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()
plt.show()
```

### 6.9.36 Why Potential Outliers Were Not Removed

The boxplots identified several observations that lie beyond the upper whiskers and may therefore be considered statistical outliers. However, these observations were **not removed automatically**.

In a pension dataset, unusually high values for income, contributions, total contributions, current fund value, pension amount, or units held may represent genuine differences between individuals rather than data-entry errors. Removing such observations solely because they fall outside the IQR-based boxplot limits could result in the loss of meaningful financial information.

Therefore, the boxplots were used as a **diagnostic tool to identify unusual observations**, rather than as a rule for deleting records. Observations were retained unless there was a clear data-quality reason to consider them invalid.

This approach preserves the original variation in the dataset and avoids introducing bias by removing legitimate high-value pension records.

# 7. Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) is the process of examining and understanding a dataset before performing detailed statistical analysis or drawing conclusions from the data.

The main purpose of EDA is to understand the structure, distribution, patterns, relationships, and potential irregularities present in the dataset. It helps identify important characteristics of the data and provides a foundation for selecting appropriate analytical methods and visualisations.

For the India Pension Scheme dataset, EDA is used to understand the demographic characteristics of pension account holders, their pension scheme participation, contribution patterns, financial outcomes, investment preferences, and relationships between important variables.

The EDA is organised into the following stages:

1. **Descriptive Analysis** – summarises the numerical characteristics of the dataset.
2. **Categorical Analysis** – examines the distribution of important categorical variables.
3. **Univariate Analysis** – studies individual variables and their distributions.
4. **Bivariate Analysis** – examines relationships between two variables.
5. **Multivariate Analysis** – examines patterns involving three or more variables.
6. **Correlation Analysis** – measures the strength and direction of relationships between numerical variables.

The analysis is primarily descriptive and exploratory. Therefore, observed patterns and relationships are interpreted as characteristics of the dataset and are not automatically treated as causal relationships.

## 7.1 Descriptive Analysis

Descriptive analysis provides a numerical summary of the dataset. It helps describe the central tendency, variability, and range of numerical variables without attempting to make predictions or establish causal relationships.

For numerical variables, measures such as the mean, median, standard deviation, minimum, maximum, and quartiles are examined. These measures help identify the typical values in the dataset, the amount of variation between observations, and the presence of unusually large or small values.

In the pension dataset, descriptive analysis is particularly useful for variables such as age, annual income, monthly contribution, total contribution, current fund value, expected monthly pension, and years of contribution.

This stage provides an initial numerical understanding of the dataset before visual analysis is performed.

## 7.4 Bivariate Analysis

Bivariate analysis examines two variables simultaneously to understand whether a relationship or pattern exists between them.

The appropriate visualisation depends on the types of variables being compared. For example, scatter plots can be used for two numerical variables, while grouped bar charts or boxplots can be used when numerical and categorical variables are compared.

For the pension dataset, bivariate analysis can be used to investigate questions such as whether higher monthly contributions are associated with higher expected monthly pensions, whether income is related to contribution levels, and whether age is associated with financial outcomes.

The relationships identified through bivariate analysis describe patterns within the dataset and should not be interpreted as proof of causation.

### 7.4.1 Annual Income vs Monthly Contribution

#### Why This Visual Is Suitable

A **hexbin plot** is suitable for examining annual income and monthly contribution because both variables are numerical and many observations overlap at similar values.

Instead of displaying every observation separately, the hexagonal areas show where account holders are more concentrated. Darker areas represent regions with more observations, making the overall relationship easier to identify.

This provides a clearer view of the data than a highly crowded scatter plot.

```python
# 7.4.1 Annual Income vs Monthly Contribution

plt.figure(figsize=(10, 6))

plt.hexbin(
    df["Annual_Income_INR"],
    df["Monthly_Contribution_INR"],
    gridsize=30,
    cmap="Purples",
    mincnt=1
)

plt.colorbar(label="Number of Account Holders")

plt.title(
    "Annual Income vs Monthly Pension Contribution",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Annual Income (INR)")
plt.ylabel("Monthly Contribution (INR)")

plt.grid(
    linestyle="--",
    alpha=0.2
)

plt.tight_layout()
plt.show()
```

### Key Observations

- The highest concentration of account holders occurs at **lower monthly contribution levels**, particularly below approximately **₹2,500**.

- As annual income increases, monthly contributions generally show an **upward association**, although the observations are widely dispersed.

- A smaller group of account holders has **substantially higher monthly contributions**, reaching above **₹10,000**.

- Higher-income account holders are distributed across a wider range of monthly contribution levels rather than following one uniform pattern.

- Overall, the visual suggests a **positive association between annual income and monthly contribution** in this dataset, while also showing considerable variation among individual account holders.

### 7.4.2 Employee vs Employer Monthly Contribution

#### Why This Visual Is Suitable

A **dumbbell chart** is suitable for comparing the typical monthly contribution made by the account holder and the employer.

The two endpoints show the median employee and employer contributions, while the connecting line highlights the difference between them.

This visual provides a direct comparison of the two contribution sources without using another bar or scatter chart.

```python
# 7.4.2 Employee vs Employer Monthly Contribution

employee_median = df["Monthly_Contribution_INR"].median()
employer_median = df["Employer_Monthly_Contribution_INR"].median()

plt.figure(figsize=(10, 4))

# Connecting line
plt.plot(
    [employer_median, employee_median],
    [0, 0],
    color="#B0BEC5",
    linewidth=6,
    zorder=1
)

# Employee contribution
plt.scatter(
    employee_median,
    0,
    s=220,
    color="#42A5F5",
    edgecolor="white",
    linewidth=2,
    zorder=2
)

# Employer contribution
plt.scatter(
    employer_median,
    0,
    s=220,
    color="#EF5350",
    edgecolor="white",
    linewidth=2,
    zorder=2
)

# Employee label
plt.annotate(
    f"Employee\n₹{employee_median:,.0f}",
    (employee_median, 0),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    fontsize=11,
    fontweight="bold"
)

# Employer label
plt.annotate(
    f"Employer\n₹{employer_median:,.0f}",
    (employer_median, 0),
    xytext=(0, -45),
    textcoords="offset points",
    ha="center",
    fontsize=11,
    fontweight="bold"
)

plt.yticks([])
plt.xlabel("Median Monthly Contribution (INR)")

plt.title(
    "Employee vs Employer Monthly Contribution",
    fontsize=16,
    fontweight="bold",
    pad=25
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

### Key Observations

- The median **employee monthly contribution is ₹3,820**.

- The median **employer monthly contribution is ₹1,250**.

- The employee contribution is substantially higher than the employer contribution based on the median values.

- The difference between the two contribution levels is clearly visible in the dumbbell chart.

- Overall, the dataset shows that **employee contributions are higher than employer contributions in terms of their median monthly values**.

## 7.6 Correlation Analysis

Correlation analysis examines the strength and direction of the relationship between numerical variables.

A correlation coefficient generally ranges from -1 to +1. A positive value indicates that two variables tend to increase together, while a negative value indicates that one variable tends to increase as the other decreases. Values closer to zero indicate a weaker linear relationship.

For the pension dataset, correlation analysis is particularly useful for examining relationships among income, contributions, total contribution, current fund value, and expected monthly pension.

Correlation analysis provides a useful summary of numerical relationships, but correlation does not establish causation. A strong correlation between two variables does not by itself demonstrate that one variable causes changes in the other.

### 7.6.1 Correlation Network of Key Pension Variables

**Why This Visual Is Suitable**

A correlation network is suitable for showing several relationships simultaneously. Each variable is represented as a node, while the lines connecting the variables represent the strength of their Spearman correlations.

- **Thicker lines** indicate stronger relationships.
- **Thinner lines** indicate weaker relationships.
- The visual allows the strongest relationships in the pension dataset to be identified quickly without creating multiple separate charts.


```python
import matplotlib.pyplot as plt
import numpy as np

# Correlation values already obtained from the analysis
correlations = {
    ("Monthly Contribution", "Current Fund Value"): 0.81,
    ("Total Contribution", "Current Fund Value"): 0.97,
    ("Total Contribution", "Expected Monthly Pension"): 0.75,
    ("Current Fund Value", "Expected Monthly Pension"): 0.76,
    ("Monthly Contribution", "Expected Monthly Pension"): 0.56,
    ("Annual Income", "Monthly Contribution"): 0.55,
    ("Years of Contribution", "Current Fund Value"): 0.43,
    ("Years of Contribution", "Expected Monthly Pension"): 0.47
}

# Variables
variables = [
    "Annual Income",
    "Monthly Contribution",
    "Years of Contribution",
    "Total Contribution",
    "Current Fund Value",
    "Expected Monthly Pension"
]

# Positions of variables
positions = {
    "Annual Income": (0, 2),
    "Monthly Contribution": (1, 2),
    "Years of Contribution": (0, 0),
    "Total Contribution": (2, 0),
    "Current Fund Value": (3, 1),
    "Expected Monthly Pension": (4, 1)
}

plt.figure(figsize=(14, 7))

# Draw connections
for (var1, var2), corr in correlations.items():

    x1, y1 = positions[var1]
    x2, y2 = positions[var2]

    linewidth = 1 + (corr * 5)

    plt.plot(
        [x1, x2],
        [y1, y2],
        linewidth=linewidth,
        alpha=0.65
    )

    # Correlation value at midpoint
    mid_x = (x1 + x2) / 2
    mid_y = (y1 + y2) / 2

    plt.text(
        mid_x,
        mid_y + 0.08,
        f"{corr:.2f}",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

# Draw variable nodes
for variable, (x, y) in positions.items():

    plt.scatter(
        x,
        y,
        s=1800,
        zorder=3,
        edgecolor="black",
        linewidth=1.2
    )

    plt.text(
        x,
        y,
        variable,
        ha="center",
        va="center",
        fontsize=9,
        fontweight="bold",
        wrap=True
    )

plt.title(
    "Correlation Network of Key Pension Variables",
    fontsize=17,
    fontweight="bold"
)

plt.xlim(-0.7, 4.7)
plt.ylim(-0.7, 2.7)

plt.axis("off")

plt.tight_layout()
plt.show()
```

**Key Observations**

- **Total Contribution and Current Fund Value** show the strongest positive association, with a Spearman correlation of **0.97**.
- **Monthly Contribution and Current Fund Value** also show a strong positive association of **0.81**.
- **Current Fund Value and Expected Monthly Pension** have a strong positive association of **0.76**.
- **Total Contribution and Expected Monthly Pension** show a strong positive association of **0.75**.
- **Annual Income and Monthly Contribution** have a moderate positive association of **0.55**.
- **Years of Contribution** has positive associations with both **Current Fund Value (0.43)** and **Expected Monthly Pension (0.47)**.
- Overall, the financial contribution variables show the strongest relationships with current fund value and expected monthly pension.
- These relationships represent **associations within the dataset and do not establish causation**.
